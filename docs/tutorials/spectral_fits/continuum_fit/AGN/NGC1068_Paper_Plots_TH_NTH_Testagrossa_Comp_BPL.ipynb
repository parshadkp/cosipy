## Cutoff Power Law (Thermal)

#### NGC 1068 (400 keV)

```python
K_inj_thermal = 3.5e-1 / u.cm / u.cm / u.s / u.keV
piv_inj_thermal = 1. * u.keV
xc_inj_thermal = 400. * u.keV
index_inj_thermal = -1.9

spectrum_inj_ec400 = Cutoff_powerlaw()

spectrum_inj_ec400.K.value = K_inj_thermal.value
spectrum_inj_ec400.piv.value = piv_inj_thermal.value
spectrum_inj_ec400.xc.value = xc_inj_thermal.value
spectrum_inj_ec400.index.value = index_inj_thermal

spectrum_inj_ec400.K.unit = K_inj_thermal.unit
spectrum_inj_ec400.piv.unit = piv_inj_thermal.unit
spectrum_inj_ec400.xc.unit = xc_inj_thermal.unit
```

## Broken Power Law (Non-thermal)

#### NGC 1068 broken-power-law tail

```python
def cutoff_powerlaw_k_at_pivot(shape, pivot_value):
    return float(shape.K.value * (pivot_value / shape.piv.value) ** shape.index.value)

pivot_energy = 1.0 * u.keV
break_energy = 640.0 * u.keV
# astromodels uses signed power-law indices: alpha=-Gamma1, beta=-Gamma2.
index_below_break = -1.9
index_above_break = -3.4
K_at_pivot = 3.5e-2 / u.cm / u.cm / u.s / u.keV

spectrum_inj_ec400_BPL = Broken_powerlaw()

spectrum_inj_ec400_BPL.K.value = K_at_pivot.value
spectrum_inj_ec400_BPL.piv.value = pivot_energy.value
spectrum_inj_ec400_BPL.xb.value = break_energy.value
spectrum_inj_ec400_BPL.alpha.value = index_below_break
spectrum_inj_ec400_BPL.beta.value = index_above_break

spectrum_inj_ec400_BPL.K.unit = K_at_pivot.unit
spectrum_inj_ec400_BPL.piv.unit = pivot_energy.unit
spectrum_inj_ec400_BPL.xb.unit = break_energy.unit

spectrum_inj_ec400_total = spectrum_inj_ec400 + spectrum_inj_ec400_BPL

fit_pivot_keV = pivot_energy.to_value(u.keV)
thermal_k_at_fit_pivot = cutoff_powerlaw_k_at_pivot(spectrum_inj_ec400, fit_pivot_keV)
tail_k_at_fit_pivot = spectrum_inj_ec400_BPL.K.value
linking_ratio_ec400 = tail_k_at_fit_pivot / thermal_k_at_fit_pivot

thermal_energy_flux, _ = quad(lambda energy: energy * spectrum_inj_ec400.evaluate_at(energy), 100.0, 10000.0)
nonthermal_energy_flux, _ = quad(lambda energy: energy * spectrum_inj_ec400_BPL.evaluate_at(energy), 100.0, 10000.0)
nonthermal_fraction = nonthermal_energy_flux / (thermal_energy_flux + nonthermal_energy_flux)

print("BPL/CPL normalization ratio at 1 keV:", linking_ratio_ec400)
print(f"Non-thermal 0.1-10 MeV energy-flux fraction: {nonthermal_fraction:.4f}")
```

## Thermal CPL + non-thermal BPL fit

```python
l=172.104
b=-51.934

# Initialize the BPL at the injected shape; fit its break and both slopes.
K_bpl = tail_k_at_fit_pivot / u.cm / u.cm / u.s / u.keV
piv_bpl = fit_pivot_keV * u.keV
xb_bpl = break_energy
alpha_bpl = index_below_break
beta_bpl = index_above_break

spectrum_bpl = Broken_powerlaw()
spectrum_bpl.K.value = K_bpl.value
spectrum_bpl.piv.value = piv_bpl.value
spectrum_bpl.xb.value = xb_bpl.value
spectrum_bpl.alpha.value = alpha_bpl
spectrum_bpl.beta.value = beta_bpl

spectrum_bpl.K.min_value = 1e-8
spectrum_bpl.K.max_value = 1.0
spectrum_bpl.xb.min_value = 100.0
spectrum_bpl.xb.max_value = 10000.0
spectrum_bpl.alpha.min_value = -5.0
spectrum_bpl.alpha.max_value = 1.0
spectrum_bpl.beta.min_value = -5.0
spectrum_bpl.beta.max_value = 1.0
spectrum_bpl.xb.delta = 20.0
spectrum_bpl.alpha.delta = 0.25
spectrum_bpl.beta.delta = 0.25

spectrum_bpl.K.unit = K_bpl.unit
spectrum_bpl.piv.unit = piv_bpl.unit
spectrum_bpl.xb.unit = xb_bpl.unit
spectrum_bpl.piv.fix = True
spectrum_bpl.xb.fix = False
spectrum_bpl.alpha.fix = False
spectrum_bpl.beta.fix = False

```

```python
from agn_cosi_fit_utils import COSIPlugin, EnergyRangeCOSIPlugin, make_cosi_background_parameter
# Keep both spectral components in one point source so the response cache
# tracks every component as linked or independently fitted parameters change.
source1 = PointSource(
    "source1",
    l=l,
    b=b,
    spectral_shape=spectrum_cpl + spectrum_bpl,
)
model = Model(source1)
cosi.set_model(model)


def copy_parameter_settings(reference_parameter, target_parameter):
    target_parameter.value = reference_parameter.value
    target_parameter.fix = reference_parameter.fix

    if reference_parameter.min_value is not None:
        target_parameter.min_value = reference_parameter.min_value
    if reference_parameter.max_value is not None:
        target_parameter.max_value = reference_parameter.max_value
    if reference_parameter.delta is not None:
        target_parameter.delta = reference_parameter.delta
    if reference_parameter.unit is not None:
        target_parameter.unit = reference_parameter.unit


def make_cpl_only_source(name, reference_source):
    reference_shape = reference_source.spectrum.main.shape.functions[0]
    cpl_shape = Cutoff_powerlaw()

    for parameter_name in ("K", "piv", "xc", "index"):
        copy_parameter_settings(
            getattr(reference_shape, parameter_name),
            getattr(cpl_shape, parameter_name),
        )
    # Let the single-component CPL optimize its own slope.
    cpl_shape.index.fix = False
    cpl_shape.index.min_value = -5.0
    cpl_shape.index.max_value = 1.0
    cpl_shape.index.delta = 0.25
    return PointSource(
        name,
        l=l,
        b=b,
        spectral_shape=cpl_shape,
    )


def make_pl_only_source(name):
    pl_shape = Powerlaw()
    pl_shape.K.value = tail_k_at_fit_pivot
    pl_shape.piv.value = fit_pivot_keV
    pl_shape.index.value = index_below_break

    pl_shape.K.min_value = 1e-8
    pl_shape.K.max_value = 1.0
    pl_shape.index.min_value = -5
    pl_shape.index.max_value = 1
    pl_shape.index.delta = 0.25

    pl_shape.K.unit = u.cm ** -2 * u.s ** -1 * u.keV ** -1
    pl_shape.piv.unit = u.keV

    return PointSource(
        name,
        l=l,
        b=b,
        spectral_shape=pl_shape,
    )


def make_bpl_only_source(name, reference_source):
    reference_shape = reference_source.spectrum.main.shape.functions[1]
    bpl_shape = Broken_powerlaw()

    for parameter_name in ("K", "piv", "xb", "alpha", "beta"):
        copy_parameter_settings(
            getattr(reference_shape, parameter_name),
            getattr(bpl_shape, parameter_name),
        )
    return PointSource(
        name,
        l=l,
        b=b,
        spectral_shape=bpl_shape,
    )


def make_cosi_plugin(name, data_hist, nuisance_name):
    nuisance_parameter = Parameter(
        nuisance_name,
        1,
        min_value=0,
        max_value=5,
        delta=0.05,
        desc=f"Background parameter for {name}",
    )

    return COSIPlugin(
        name,
        dr=dr,
        data=data_hist.project("Em", "Phi", "PsiChi"),
        bkg=bkg.project("Em", "Phi", "PsiChi"),
        sc_orientation=sc_orientation,
        nuisance_param=nuisance_parameter,
        earth_occ=True,
    )


source1_cpl_only = make_cpl_only_source("source1_cpl_only", source1)
source1_pl_only = make_pl_only_source("source1_pl_only")
source1_bpl_only = make_bpl_only_source("source1_bpl_only", source1)

model_cpl_only = Model(source1_cpl_only)
model_pl_only = Model(source1_pl_only)
model_bpl_only = Model(source1_bpl_only)

cosi_cpl_only = make_cosi_plugin(
    "cosi_cpl_only",
    NGC1068_ec400_bkg,
    "background_cosi_cpl_only",
)
cosi_pl_only = make_cosi_plugin(
    "cosi_pl_only",
    NGC1068_ec400_bkg,
    "background_cosi_pl_only",
)
cosi_bpl_only = make_cosi_plugin(
    "cosi_bpl_only",
    NGC1068_ec400_bkg,
    "background_cosi_bpl_only",
)

cosi_cpl_only.set_model(model_cpl_only)
cosi_pl_only.set_model(model_pl_only)
cosi_bpl_only.set_model(model_bpl_only)
```

### CPL + BPL vs PL, CPL, and BPL comparisons

```python
from agn_cosi_fit_utils import COSIPlugin, EnergyRangeCOSIPlugin, make_cosi_background_parameter
def make_null_likelihood(data_hist, label):
    bkg_par_null = Parameter(
        f"background_cosi_null_{label}",
        1,
        min_value=0,
        max_value=5,
        delta=0.05,
        desc="Background parameter for the null COSI fit",
    )

    cosi_null = COSIPlugin(
        f"cosi_null_{label}",
        dr=dr,
        data=data_hist.project("Em", "Phi", "PsiChi"),
        bkg=bkg.project("Em", "Phi", "PsiChi"),
        sc_orientation=sc_orientation,
        nuisance_param=bkg_par_null,
        earth_occ=True,
    )

    spectrum_null = Powerlaw()
    spectrum_null.K.value = 1e-30
    spectrum_null.index.value = 1
    spectrum_null.K.fix = True
    spectrum_null.index.fix = True

    source_null = PointSource(
        "source_null",
        l=l,
        b=b,
        spectral_shape=spectrum_null,
    )

    model_null = Model(source_null)
    cosi_null.set_model(model_null)

    plugins_null = DataList(cosi_null)
    like_null = JointLikelihood(model_null, plugins_null, verbose=False)
    like_null.fit()

    return like_null


def get_likelihood_statistic(joint_likelihood):
    statistic_frame = joint_likelihood.results.get_statistic_frame()
    statistic_series = statistic_frame["-log(likelihood)"]

    if "total" in statistic_series.index:
        return float(statistic_series.loc["total"])

    return float(statistic_series.sum())


def detection_ts(null_likelihood, source_likelihood):
    return 2.0 * (
        get_likelihood_statistic(null_likelihood)
        - get_likelihood_statistic(source_likelihood)
    )


def model_delta_ts(reference_likelihood, test_likelihood):
    return 2.0 * (
        get_likelihood_statistic(reference_likelihood)
        - get_likelihood_statistic(test_likelihood)
    )


like_null_400 = make_null_likelihood(NGC1068_ec400_bkg, "ec400_cpl_bpl")

TS_cpl_bpl_400 = detection_ts(like_null_400, like)
TS_pl_400 = detection_ts(like_null_400, like_pl_only)
TS_cpl_400 = detection_ts(like_null_400, like_cpl_only)
TS_bpl_400 = detection_ts(like_null_400, like_bpl_only)
delta_TS_cpl_bpl_vs_pl_400 = model_delta_ts(like_pl_only, like)
delta_TS_cpl_bpl_vs_cpl_400 = model_delta_ts(like_cpl_only, like)
delta_TS_cpl_bpl_vs_bpl_400 = model_delta_ts(like_bpl_only, like)

fit_ts_comparison = pd.DataFrame(
    [
        {
            "spectrum": "NGC 1068 Ec=400 keV; CPL + BPL injection",
            "TS_CPL_plus_BPL": TS_cpl_bpl_400,
            "TS_PL": TS_pl_400,
            "TS_CPL": TS_cpl_400,
            "TS_BPL": TS_bpl_400,
            "Delta_TS_CPL_plus_BPL_vs_PL": delta_TS_cpl_bpl_vs_pl_400,
            "Delta_TS_CPL_plus_BPL_vs_CPL": delta_TS_cpl_bpl_vs_cpl_400,
            "Delta_TS_CPL_plus_BPL_vs_BPL": delta_TS_cpl_bpl_vs_bpl_400,
            "Sigma_CPL_plus_BPL": np.sqrt(max(TS_cpl_bpl_400, 0.0)),
            "Sigma_PL": np.sqrt(max(TS_pl_400, 0.0)),
            "Sigma_CPL": np.sqrt(max(TS_cpl_400, 0.0)),
            "Sigma_BPL": np.sqrt(max(TS_bpl_400, 0.0)),
        },
    ]
)

display(fit_ts_comparison)

for _, row in fit_ts_comparison.iterrows():
    print(row["spectrum"])
    print(f"  CPL + BPL TS: {row['TS_CPL_plus_BPL']:.3f} ({row['Sigma_CPL_plus_BPL']:.2f} sigma)")
    print(f"  PL TS: {row['TS_PL']:.3f} ({row['Sigma_PL']:.2f} sigma)")
    print(f"  CPL TS: {row['TS_CPL']:.3f} ({row['Sigma_CPL']:.2f} sigma)")
    print(f"  BPL TS: {row['TS_BPL']:.3f} ({row['Sigma_BPL']:.2f} sigma)")
    print(f"  CPL + BPL vs PL dTS: {row['Delta_TS_CPL_plus_BPL_vs_PL']:.3f}")
    print(f"  CPL + BPL vs CPL dTS: {row['Delta_TS_CPL_plus_BPL_vs_CPL']:.3f}")
    print(f"  CPL + BPL vs BPL dTS: {row['Delta_TS_CPL_plus_BPL_vs_BPL']:.3f}")

# Backward-compatible alias used by the plotting cells below.
TS = TS_cpl_bpl_400
```
