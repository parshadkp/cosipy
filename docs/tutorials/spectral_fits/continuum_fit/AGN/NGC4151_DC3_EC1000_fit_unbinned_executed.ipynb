# Fitting the DC3 NGC 4151 EC1000 Spectrum with a Cutoff Power Law

## Introduction

This notebook demonstrates an unbinned spectral analysis of the DC3 NGC 4151 EC1000 simulation using a cutoff power-law (CPL) spectrum. Only the source-event file is from DC3; the background events, orientation, neural-network response, and neural-network background model are the DC4 files used by `NGC4151_fit_unbinned.ipynb`.

> **Note on Underlying Mechanics:** To keep this tutorial concise, the detailed explanations of the underlying classes (such as `NFResponse`, `NFBackground`, and `UnbinnedThreeMLPointSourceResponseIRFAdaptive`) have been omitted here. If you are unfamiliar with the Neural-Network Response, the Background Approximation, or the corresponding hardware requirements introduced for DC4, **please refer to the foundational tutorial:** `example_grb_fit_normalizing_flows.ipynb`.

### Initializing Models and Folding Objects
Setting up the neural-network response (`NFResponse`), the background approximation (`NFBackground`), and the adaptive folding objects.

```python
import torch

if not torch.cuda.is_available():
    raise RuntimeError(
        "No CUDA GPU is available. Run this notebook inside a GPU Slurm job."
    )

device_name = os.environ.get("COSI_DEVICE", "cuda:0")
device = torch.device(device_name)
if device.type != "cuda" or device.index is None:
    raise ValueError("COSI_DEVICE must be an indexed CUDA device, such as cuda:0")
if device.index >= torch.cuda.device_count():
    raise RuntimeError(
        f"Requested {device_name}, but only {torch.cuda.device_count()} GPU(s) are visible"
    )

devices = [device_name]
print(f"Visible GPUs: {torch.cuda.device_count()}")
print(f"Using {device_name}: {torch.cuda.get_device_name(device)}")
print(f"PyTorch: {torch.__version__}; CUDA: {torch.version.cuda}")

log_stage(f"Using {device_name}: {torch.cuda.get_device_name(device)}")
```

Use the PyTorch cutoff power law with a shape correction for the expected-count integral. The response subclass folds the cached weights on CUDA without returning every energy-node flux to the CPU. Only the reduced density for each event is copied back for the existing likelihood. Progress and likelihood timings are saved to `NGC4151-dc3-ec1000-cpl.log` in the working directory.

```python
class CutoffPowerlawPyTorch(FastCutoffPowerlawPyTorch):
    # Reuse the existing astromodels parameter and device definitions.
    __doc__ = FastCutoffPowerlawPyTorch.__doc__
    _set_units = FastCutoffPowerlawPyTorch._set_units

    def evaluate(self, x, K, piv, index, xc):
        # Preserve the energy shape for scalar fit parameters. The base
        # implementation returns (N, 1), which broadcasts incorrectly in
        # the expected-count integral. Array parameters are used by
        # results.propagate(); allow their shape to broadcast out of place.
        with_units = isinstance(x, u.Quantity)
        energy = x.value if with_units else x
        values = [value.value if isinstance(value, u.Quantity) else value
                  for value in (K, piv, index, xc)]
        result = self.evaluate_tensor(energy, *values).cpu().numpy()
        return result * self.y_unit if with_units else result

    def evaluate_tensor(self, energy_keV, K=None, piv=None, index=None, xc=None):
        """Evaluate on torch without returning energy-node fluxes to NumPy."""
        energy = torch.as_tensor(
            energy_keV, dtype=torch.float64, device=self.devices.value)
        values = [self.K.value if K is None else K,
                  self.piv.value if piv is None else piv,
                  self.index.value if index is None else index,
                  self.xc.value if xc is None else xc]
        if any(np.ndim(value) > 0 for value in values):
            energy = energy.reshape(-1, 1)
            values = [torch.as_tensor(value, dtype=torch.float64,
                                      device=self.devices.value)
                      for value in values]
        K, piv, index, xc = values
        return K * (energy / piv).pow(index) * torch.exp(-energy / xc)


class TorchCutoffPointSourceResponse(UnbinnedThreeMLPointSourceResponseIRFAdaptive):
    """Fold the cached response with the cutoff spectrum on its torch device."""

    def cache_from_file(self, filename):
        # HDF5 loading restores the old batch size. Keep this run's setting.
        integration_batch_size = self.integration_batch_size
        super().cache_from_file(filename)
        self.integration_batch_size = integration_batch_size
        self.force_energy_node_caching = True

    @torch.inference_mode()
    def expectation_density(self):
        self._update_cache()
        spectrum = self._source.spectrum.main.shape
        if not isinstance(spectrum, CutoffPowerlawPyTorch):
            raise TypeError("Torch folding requires CutoffPowerlawPyTorch")
        source_dict = self._source.to_dict()

        if (source_dict != self._last_convolved_source_dict_density
                or self._exp_density is None):
            if self._irf_energy_node_cache is None:
                raise RuntimeError("Torch folding requires cached energy nodes")
            self._exp_density = torch.empty(self._n_events, dtype=torch.float64)
            n_energy = self._total_energy_nodes[0]
            batch_size_events = self.integration_batch_size // n_energy

            for start in range(0, self._n_events, batch_size_events):
                stop = min(start + batch_size_events, self._n_events)
                nodes = torch.as_tensor(
                    self._irf_energy_node_cache[start:stop],
                    dtype=torch.float64, device=spectrum.devices.value)
                flux = spectrum.evaluate_tensor(nodes)
                weights = self._irf_cache[start:stop].to(
                    dtype=torch.float64, device=spectrum.devices.value)
                density = torch.linalg.vecdot(weights, flux, dim=1)
                self._exp_density[start:stop].copy_(density.cpu())

            self._last_convolved_source_dict_density = source_dict

        result = self._exp_density.numpy()
        return result if self._offset is None else result + self._offset


class TimedUnbinnedLikelihood(UnbinnedLikelihood):
    """Record each evaluation even if nbconvert never writes its notebook."""

    def __init__(self, expectation):
        super().__init__(expectation)
        self.evaluation_count = 0

    def get_log_like(self):
        self.evaluation_count += 1
        count = self.evaluation_count
        started = perf_counter()
        run_logger.info("Likelihood evaluation %d started", count)
        value = super().get_log_like()
        run_logger.info(
            "Likelihood evaluation %d finished in %.2fs; logL=%.12g",
            count, perf_counter() - started, value)
        return value
```

```python
rsp = NFResponse(
    path_to_model=rsp_path,
    area_batch_size=300_000,
    density_batch_size=100_000, 
    devices=devices,
    area_compile_mode=None,
    density_compile_mode=None,
    show_progress=True)

irf = UnpolarizedNFFarFieldInstrumentResponseFunction(rsp)
```

```python
bkg_model = NFBackground(
    path_to_model=bkg_path,
    density_batch_size=100_000,
    devices=devices,
    density_compile_mode=None,
    show_progress=True)

bkg = FreeNormNFUnbinnedBackground(
    model=bkg_model, 
    data=data, 
    sc_history=sc_orientation, 
    label="bkg_norm")
```

```python
log_stage("Preparing event coordinates for the response")
psr = TorchCutoffPointSourceResponse(
    data=data,
    irf=irf,
    sc_history=sc_orientation,
    show_progress=True,
    force_energy_node_caching=True,
    reduce_memory=True)

psr.cache_batch_size = 100_000
# This counts energy nodes: 3,000,000 / 60 = 50,000 events per batch.
# Only a batch is transferred to CUDA; the full response stays in host RAM.
psr.integration_batch_size = 3_000_000
log_stage("Response coordinate preparation complete")
```

The DC3 EC1000 source was generated with a single `Cutoff_powerlaw`: normalization 0.15 keV$^{-1}$ s$^{-1}$ cm$^{-2}$ at 1 keV, photon index -1.75, and cutoff energy 1000 keV. All three parameters remain free in the fit. The numerical pivot is moved to 100 keV while preserving the injected spectrum exactly.

The fitted spectrum uses `CutoffPowerlawPyTorch` on the selected CUDA device. Cached response weights and energy nodes are reused, and spectral folding runs in bounded GPU batches with float64 arithmetic. The expected-count integral and the threeML likelihood keep their existing interfaces.

```python
l = 155.07
b = 75.06

index = -1.75  # Injected photon index
piv = 100. * u.keV  # Pivot inside the fitted band for better conditioning
xc = 1000. * u.keV
K_1keV = 0.15 / u.cm / u.cm / u.s / u.keV
K = K_1keV * (piv / (1. * u.keV))**index

spectrum = CutoffPowerlawPyTorch(
    index=index,
    K=K.value,
    piv=piv.value,
    xc=xc.value,
)

spectrum.index.min_value = -5
spectrum.index.max_value = 0

spectrum.xc.min_value = 50
spectrum.xc.max_value = 5000

spectrum.K.min_value = K.value / 100
spectrum.K.max_value = K.value * 100

spectrum.K.unit = K.unit
spectrum.piv.unit = piv.unit
spectrum.xc.unit = xc.unit

spectrum.index.delta = 0.01
spectrum.K.delta = K.value * 0.1
spectrum.xc.delta = 50.

spectrum.devices.value = device_name
```

```python
# Reference implementation of the exact injected spectrum, evaluated on CPU.
spectrum_inj = Cutoff_powerlaw(
    index=index,
    K=K.value,
    piv=piv.value,
    xc=xc.value,
)
```

```python
source = PointSource("NGC_4151",
                     l=l,
                     b=b,
                     spectral_shape=spectrum)
model = Model(source)
```

```python
response = CachedUnbinnedThreeMLModelFolding(psr)
```

```python
expectation_density = SumExpectationDensity(response, bkg)
```

```python
like_fun = TimedUnbinnedLikelihood(expectation_density)
cosi = ThreeMLPluginInterface('cosi', like_fun, response, bkg)
```

```python
bkg_norm = bkg.norm.to_value(u.Hz)

cosi.bkg_parameter['bkg_norm'] = Parameter("bkg_norm",
                                          bkg_norm,
                                          unit = u.Hz,
                                          min_value=0,
                                          max_value=150,
                                          delta=0.05,
                                          )

print(f"The average background flux is {bkg_norm:.2f} Hz")
```

```python
plugins = DataList(cosi)
like = JointLikelihood(model, plugins, verbose=False) # You can enable debugging
```

Now we can compare the fitted CPL with the injected DC3 EC1000 spectrum.

```python
results = like.results

parameters = {par.name: results.get_variates(par.path)
              for par in results.optimized_model["NGC_4151"].parameters.values()
              if par.free}

results_err = results.propagate(results.optimized_model["NGC_4151"].spectrum.main.shape.evaluate_at, **parameters)
```

```python
energy = np.geomspace(100*u.keV, 10*u.MeV).to_value(u.keV)

flux_lo = np.zeros_like(energy)
flux_median = np.zeros_like(energy)
flux_hi = np.zeros_like(energy)
flux_inj = np.zeros_like(energy)

for i, e in enumerate(energy):
    flux = results_err(e)
    flux_median[i] = flux.median
    flux_lo[i], flux_hi[i] = flux.equal_tail_interval(cl=0.68)
    flux_inj[i] = spectrum_inj.evaluate_at(e)
```

```python
%matplotlib inline
```

```python
fig, ax = plt.subplots(figsize = (9, 6))

ax.plot(energy, flux_median, label="Best-fit CPL")
ax.fill_between(energy, flux_lo, flux_hi, alpha=.35, label="Best-fit CPL (errors)")
ax.plot(energy, flux_inj, color="black", ls=":", lw=2,
        label="Injected CPL (cutoff = 1000 keV)")

ax.semilogx()
ax.semilogy()

ax.set_xlabel("Energy [keV]")
ax.set_ylabel(r"$\frac{\mathrm{d}N}{\mathrm{d}E}$ [keV$^{-1}$ cm$^{-2}$ s$^{-1}$]")

ax.legend()

plot_path = Path(
    "/home/parshap/cosipy/docs/tutorials/"
    "spectral_fits/continuum_fit/AGN/NGC4151_DC3_EC1000_CPL_fit.pdf"
)
fig.savefig(
    plot_path,
    bbox_inches="tight",
    facecolor="white"
)

log_stage(f"Saved spectrum plot: {plot_path}")
```
