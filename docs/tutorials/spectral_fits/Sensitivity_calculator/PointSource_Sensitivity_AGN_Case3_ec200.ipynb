# NGC 4151 Case 3: CPL + CPL, Ec=200 keV

Runs 100 Poisson realizations at 24 months using the fixed CPL + CPL normalization from the point-source injector. It saves one representative realization selected closest to the median added-component Delta TS, with the median recovered `norm_nt` used as the equal-distance tie-breaker.


```python
from IPython.display import display
from agn_fixed_case_sensitivity import (
    load_sed_handoff,
    plot_fixed_case_ensemble,
    run_fixed_case_ensemble,
    save_fixed_case_companion_exposure,
)
%matplotlib inline
```

```python
case = {
    "case_tag": "Case3_ec200",
    "output_prefix": "NGC4151_ec200_CPL_CPL",
    "description": "Case 3: thermal CPL Ec=200 keV plus CPL Ec=2000 keV",
    "file_stem": "ec_200_DC4_COSI_cpl_cpl_60_fovCut",
    "comparison": "added_component",
    "longitude_deg": 155.077, "latitude_deg": 75.063, "fov_cut_deg": 60.0,
    "fit_pivot_keV": 1.0,
    "primary": {"kind": "cpl", "K": 0.15, "pivot_keV": 1.0, "cutoff_keV": 200.0, "index": -1.75, "fit_index_free": False, "fit_cutoff_free": True},
    # Match Point_source_injector_NGC4151.ipynb: K_secondary = 0.10 * thermal flux density at 1 keV.
    "secondary": {"kind": "cpl", "pivot_keV": 1.0, "cutoff_keV": 2000.0, "index": -1.7, "fit_index_free": True, "fit_cutoff_free": True, "normalization": {"mode": "flux_ratio", "ratio": 0.10, "energy_keV": 1.0}},
}
exposure_months = 24
n_seeds = 300
seed_start = 4100
target_delta_ts = 9.0
resume_existing_trials = True
```

```python
print(f"=== {exposure_months}-month fixed-normalization ensemble ===")
result = run_fixed_case_ensemble(
    case, n_seeds=n_seeds, seed_start=seed_start,
    exposure_months=exposure_months, target_delta_ts=target_delta_ts,
    resume_existing_trials=resume_existing_trials,
)
display(result["summary"])
display(result["delta_ts_candidates"][[
    "seed", "delta_ts", "delta_ts_distance_from_median",
    "fit_secondary_link_ratio",
    "recovered_norm_nt_distance_from_median",
]])

# Save a 3-month realization at the same injected normalization and using
# the representative seed selected by the 24-month 300-seed ensemble.
companion_3m_from_24m = save_fixed_case_companion_exposure(
    result, exposure_months=3,
)
```

```python
fig, ax = plot_fixed_case_ensemble(result)
ax.set_title(f"{case['description']}, {exposure_months} months")
```

```python
sed_handoff = load_sed_handoff(result["manifest_file"])
print(exposure_months, sed_handoff[0]["median_data_file"])
companion_handoff_3m = load_sed_handoff(
    companion_3m_from_24m["manifest_file"]
)
print(3, companion_handoff_3m[0]["median_data_file"])
```
