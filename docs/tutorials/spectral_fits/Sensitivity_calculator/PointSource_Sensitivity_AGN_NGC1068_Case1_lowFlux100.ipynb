# NGC 1068 low-flux Case 1 and Testagrossa-reference sensitivity

The thermal Case 1 normalization is divided by 100 (K = 0.0031 at 1 keV). A second 300-seed reference reproduces `NGC1068_Paper_Plots_NTH_Testagrossa.ipynb` (K = 0.00031, index = -1.85), so the later side-by-side fit uses fluctuated ensembles for both spectra.


```python
from IPython.display import display
from agn_fixed_case_sensitivity import (
    load_sed_handoff,
    plot_fixed_case_ensemble,
    run_fixed_case_ensemble,
)
%matplotlib inline
```

```python
RESPONSE_PATH = (
    "/Users/parshadkp/Library/CloudStorage/OneDrive-ClemsonUniversity/"
    "COSI/Radio_Quiet_AGN/DC4_Files/"
    "ResponseContinuum.o3.e100_10000.b10log.s10396905069491.m2284."
    "filtered.nonsparse.binnedimaging.imagingresponse.h5"
)
ORIENTATION_PATH = (
    "/Users/parshadkp/Library/CloudStorage/OneDrive-ClemsonUniversity/"
    "COSI/Radio_Quiet_AGN/DC4_Files/"
    "DC4_final_530km_3_month_with_slew_15sbins_GalacticEarth_SAA.fits"
)
BACKGROUND_PATH = (
    "/Users/parshadkp/Library/CloudStorage/OneDrive-ClemsonUniversity/"
    "COSI/Radio_Quiet_AGN/DC4_Files/Background/"
    "Total_DC4_BG_3months_binned_data_filtered_with_SAAcut_withSAAbck_"
    "NGC1068_60deg_fov_cut.hdf5"
)

LOW_FLUX_FACTOR = 100.0
case = {
    "source_name": "NGC1068",
    "case_tag": "NGC1068_Case1_lowFlux100",
    "output_prefix": "NGC1068_Case1_lowFlux100_CPL",
    "description": "NGC 1068 Case 1: thermal CPL at 1/100 standard normalization",
    "file_stem": "ec_128_DC4_COSI_cpl_60_fovCut",
    "file_detail": "_normTH_0p0031",
    "comparison": "source_detection",
    "longitude_deg": 172.104,
    "latitude_deg": -51.934,
    "fov_cut_deg": 60.0,
    "response_path": RESPONSE_PATH,
    "orientation_path": ORIENTATION_PATH,
    "background_path": BACKGROUND_PATH,
    "fit_pivot_keV": 200.0,
    "primary": {
        "kind": "cpl", "K": 3.1e-1 / LOW_FLUX_FACTOR,
        "pivot_keV": 1.0, "cutoff_keV": 128.0, "index": -2.10,
        "fit_index_free": False, "fit_cutoff_free": True,
        "cutoff_min_keV": 10.0, "cutoff_max_keV": 2000.0,
    },
}


testagrossa_reference_case = {
    **case,
    "case_tag": "NGC1068_TestagrossaReference",
    "output_prefix": "NGC1068_TestagrossaReference_CPL",
    "description": "NGC 1068 Testagrossa-reference CPL at 1/1000 standard normalization",
    "file_detail": "_normTH_0p00031_index_m1p85",
    "primary": {
        **case["primary"],
        "K": 3.1e-1 / 1000.0,
        "index": -1.85,
        "fit_index_free": True,
        "cutoff_max_keV": 5000.0,
    },
}

exposure_months_list = (3, 24)
n_seeds = 300
seed_start = 116800
target_delta_ts = 9.0
resume_existing_trials = False
```

```python
low_flux_results = {}
testagrossa_reference_results = {}
for exposure_months in exposure_months_list:
    print(f"\n=== {exposure_months}-month 1/100 thermal ensemble ===")
    low_flux_results[exposure_months] = run_fixed_case_ensemble(
        case,
        n_seeds=n_seeds,
        seed_start=seed_start,
        exposure_months=exposure_months,
        target_delta_ts=target_delta_ts,
        resume_existing_trials=resume_existing_trials,
    )
    display(low_flux_results[exposure_months]["summary"])

    print(f"\n=== {exposure_months}-month Testagrossa-reference ensemble ===")
    testagrossa_reference_results[exposure_months] = run_fixed_case_ensemble(
        testagrossa_reference_case,
        n_seeds=n_seeds,
        seed_start=seed_start,
        exposure_months=exposure_months,
        target_delta_ts=target_delta_ts,
        resume_existing_trials=resume_existing_trials,
    )
    display(testagrossa_reference_results[exposure_months]["summary"])
```

```python
for exposure_months in exposure_months_list:
    fig, ax = plot_fixed_case_ensemble(low_flux_results[exposure_months])
    ax.set_title(f"1/100 thermal CPL, {exposure_months} months")
    fig, ax = plot_fixed_case_ensemble(testagrossa_reference_results[exposure_months])
    ax.set_title(f"Testagrossa-reference CPL, {exposure_months} months")
```

```python
low_flux_sed_handoffs = {}
testagrossa_sed_handoffs = {}
for exposure_months in exposure_months_list:
    low_flux_sed_handoffs[exposure_months] = load_sed_handoff(
        low_flux_results[exposure_months]["manifest_file"]
    )
    testagrossa_sed_handoffs[exposure_months] = load_sed_handoff(
        testagrossa_reference_results[exposure_months]["manifest_file"]
    )
    print("low flux", exposure_months, low_flux_sed_handoffs[exposure_months][0]["median_data_file"])
    print("Testagrossa", exposure_months, testagrossa_sed_handoffs[exposure_months][0]["median_data_file"])
```
