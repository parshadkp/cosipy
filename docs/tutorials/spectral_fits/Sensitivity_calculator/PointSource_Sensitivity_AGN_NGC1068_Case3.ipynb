# NGC 1068 Case 3 sensitivity: CPL + CPL

Runs 300 Poisson realizations at the injected Testagrossa normalization for 24 months. The representative file is selected by median added-component ΔTS, then median recovered link normalization. A matched 3-month companion is also saved.


```python
from IPython.display import display
from agn_fixed_case_sensitivity import (
    load_sed_handoff,
    plot_fixed_case_ensemble,
    run_fixed_case_ensemble,
    save_fixed_case_companion_exposure,
)
%matplotlib inline
```

```python
RESPONSE_PATH = (
    "/Users/parshadkp/Library/CloudStorage/OneDrive-ClemsonUniversity/"
    "COSI/Radio_Quiet_AGN/DC4_Files/"
    "ResponseContinuum.o3.e100_10000.b10log.s10396905069491.m2284."
    "filtered.nonsparse.binnedimaging.imagingresponse.h5"
)
ORIENTATION_PATH = (
    "/Users/parshadkp/Library/CloudStorage/OneDrive-ClemsonUniversity/"
    "COSI/Radio_Quiet_AGN/DC4_Files/"
    "DC4_final_530km_3_month_with_slew_15sbins_GalacticEarth_SAA.fits"
)
BACKGROUND_PATH = (
    "/Users/parshadkp/Library/CloudStorage/OneDrive-ClemsonUniversity/"
    "COSI/Radio_Quiet_AGN/DC4_Files/Background/"
    "Total_DC4_BG_3months_binned_data_filtered_with_SAAcut_withSAAbck_"
    "NGC1068_60deg_fov_cut.hdf5"
)

case = {
    "source_name": "NGC1068",
    "case_tag": "NGC1068_Case3",
    "output_prefix": "NGC1068_Case3_CPL_CPL",
    "description": "NGC 1068 Case 3: thermal CPL Ec=128 keV plus CPL Ec=3750 keV",
    "file_stem": "ec_128_DC4_COSI_cpl_cpl_60_fovCut",
    "comparison": "added_component",
    "longitude_deg": 172.104,
    "latitude_deg": -51.934,
    "fov_cut_deg": 60.0,
    "response_path": RESPONSE_PATH,
    "orientation_path": ORIENTATION_PATH,
    "background_path": BACKGROUND_PATH,
    "fit_pivot_keV": 200.0,
    "primary": {
        "kind": "cpl", "K": 3.5e-1, "pivot_keV": 1.0,
        "cutoff_keV": 128.0, "index": -1.9,
        "fit_index_free": False, "fit_cutoff_free": True,
        "cutoff_min_keV": 10.0, "cutoff_max_keV": 2000.0,
    },
    "secondary": {
        "kind": "cpl", "pivot_keV": 1.0,
        "cutoff_keV": 3750.0, "index": -1.85,
        "fit_index_free": True, "fit_cutoff_free": True,
        "cutoff_min_keV": 1000.0, "cutoff_max_keV": 10000.0,
        "normalization": {
            "mode": "flux_ratio", "ratio": 0.10, "energy_keV": 1.0,
        },
    },
}
exposure_months = 24
n_seeds = 300
seed_start = 106800
target_delta_ts = 9.0
resume_existing_trials = True
```

```python
print(f"=== {exposure_months}-month fixed-normalization ensemble ===")
result = run_fixed_case_ensemble(
    case,
    n_seeds=n_seeds,
    seed_start=seed_start,
    exposure_months=exposure_months,
    target_delta_ts=target_delta_ts,
    resume_existing_trials=resume_existing_trials,
)
display(result["summary"])
display(result["delta_ts_candidates"][[
    "seed", "delta_ts", "delta_ts_distance_from_median",
    "fit_secondary_link_ratio",
    "recovered_norm_nt_distance_from_median",
]])

companion_3m_from_24m = save_fixed_case_companion_exposure(
    result,
    exposure_months=3,
)
```

```python
fig, ax = plot_fixed_case_ensemble(result)
ax.set_title(f"{case['description']}, {exposure_months} months")
```

```python
sed_handoff_24m = load_sed_handoff(result["manifest_file"])
sed_handoff_3m = load_sed_handoff(companion_3m_from_24m["manifest_file"])
print(24, sed_handoff_24m[0]["median_data_file"])
print(3, sed_handoff_3m[0]["median_data_file"])
```
