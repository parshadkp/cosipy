# NGC 4151 Case 2 threshold: CPL + PL, Ec=200 keV

Scans the PL-to-thermal flux ratio `norm_nt` at 24 months and finds the lowest tested normalization whose median added-tail Delta TS reaches 9. At that normalization, it selects the realization closest to median Delta TS; equal-distance candidates are ranked by closeness to median recovered `norm_nt`, then by seed. It also saves a 3-month companion using the selected spectrum and seed. No 3-month Delta-TS ensemble is run.


```python
import numpy as np
from IPython.display import display
from agn_fixed_case_sensitivity import (
    load_sed_handoff,
    plot_norm_threshold_ensemble,
    run_norm_threshold_ensemble,
    save_threshold_companion_exposure,
)
%matplotlib inline
```

```python
case = {
    "case_tag": "Case2_ec200",
    "output_prefix": "NGC4151_ec200_NTH_threshold",
    "description": "Case 2: thermal CPL Ec=200 keV plus PL tail",
    "file_stem": "ec_200_DC4_COSI_cpl_pl_60_fovCut",
    "comparison": "added_component",
    "longitude_deg": 155.077, "latitude_deg": 75.063, "fov_cut_deg": 60.0,
    "fit_pivot_keV": 200.0,
    "primary": {"kind": "cpl", "K": 0.15, "pivot_keV": 1.0, "cutoff_keV": 200.0, "index": -1.75, "fit_index_free": False, "fit_cutoff_free": True},
    "secondary": {"kind": "pl", "pivot_keV": 200.0, "index": -2.8, "fit_index_free": True, "normalization": {"mode": "flux_ratio", "ratio": 0.65, "energy_keV": 200.0}},
}
exposure_months = 24
norm_nt_grid = np.array([0.17])
n_seeds = 300
seed_start = 4100
target_delta_ts = 9.0
resume_existing_trials = True
```

```python
print(f"=== {exposure_months}-month threshold scan ===")
threshold_result = run_norm_threshold_ensemble(
    case, norm_nt_grid, n_seeds=n_seeds,
    seed_start=seed_start, exposure_months=exposure_months,
    target_delta_ts=target_delta_ts,
    resume_existing_trials=resume_existing_trials,
)
display(threshold_result["threshold_summary"])
display(threshold_result["delta_ts_candidates"][[
    "seed", "delta_ts", "delta_ts_distance_from_median",
    "fit_secondary_link_ratio",
    "recovered_norm_nt_distance_from_median",
]])

# Save a 3-month companion at the 24-month-selected norm_nt and median seed.
# This rescales the exposure and draws the corresponding 3-month data, but
# it does not run a separate 3-month Delta-TS ensemble.
companion_3m_result = save_threshold_companion_exposure(
    threshold_result,
    exposure_months=3,
)
```

```python
display(threshold_result["threshold_summary"])
display(threshold_result["delta_ts_candidates"][[
    "seed", "delta_ts", "delta_ts_distance_from_median",
    "fit_secondary_link_ratio",
    "recovered_norm_nt_distance_from_median",
]])
```

```python
fig, axes = plot_norm_threshold_ensemble(threshold_result)
fig.suptitle(f"{case['description']}, {exposure_months} months")
```

```python
sed_handoff_24m = load_sed_handoff(threshold_result["manifest_file"])
sed_handoff_3m = load_sed_handoff(companion_3m_result["manifest_file"])
print(24, sed_handoff_24m[0]["median_data_file"])
print(3, sed_handoff_3m[0]["median_data_file"])
```
