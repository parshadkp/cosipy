# NGC 4151 Case 2 threshold: CPL + PL, Ec=1000 keV

Scans the PL-to-thermal flux ratio `norm_nt` with 300 seeds at both 3 and 24 months. At each threshold normalization, it selects the realization closest to median Delta TS; equal-distance candidates are ranked by closeness to median recovered `norm_nt`, then by seed. The native 3-month threshold realization is preserved, and an additional 3-month companion is saved at the 24-month-selected `norm_nt`.


```python
import numpy as np
from IPython.display import display
from agn_fixed_case_sensitivity import (
    load_sed_handoff,
    plot_norm_threshold_ensemble,
    run_norm_threshold_ensemble,
    save_threshold_companion_exposure,
)
%matplotlib inline
```

```python
case = {
    "case_tag": "Case2_ec1000",
    "output_prefix": "NGC4151_ec1000_NTH_threshold",
    "description": "Case 2: thermal CPL Ec=1000 keV plus PL tail",
    "file_stem": "ec_1000_DC4_COSI_cpl_pl_60_fovCut",
    "comparison": "added_component",
    "longitude_deg": 155.077, "latitude_deg": 75.063, "fov_cut_deg": 60.0,
    "fit_pivot_keV": 200.0,
    "primary": {"kind": "cpl", "K": 0.15, "pivot_keV": 1.0, "cutoff_keV": 1000.0, "index": -1.75, "fit_index_free": False, "fit_cutoff_free": True},
    "secondary": {"kind": "pl", "pivot_keV": 200.0, "index": -2.8, "fit_index_free": True, "normalization": {"mode": "flux_ratio", "ratio": 0.65, "energy_keV": 200.0}},
}
exposure_months_list = (3, 24)
norm_nt_grids = {
    3: np.array([0.36]),
    24: np.array([0.12]),
}
n_seeds = 300
seed_start = 4100
target_delta_ts = 9.0
resume_existing_trials = True
```

```python
threshold_results = {}
for exposure_months in exposure_months_list:
    print(f"\n=== {exposure_months}-month threshold scan ===")
    threshold_results[exposure_months] = run_norm_threshold_ensemble(
        case, norm_nt_grids[exposure_months], n_seeds=n_seeds,
        seed_start=seed_start, exposure_months=exposure_months,
        target_delta_ts=target_delta_ts,
        resume_existing_trials=resume_existing_trials,
    )
    display(threshold_results[exposure_months]["threshold_summary"])
    display(threshold_results[exposure_months]["delta_ts_candidates"][[
        "seed", "delta_ts", "delta_ts_distance_from_median",
        "fit_secondary_link_ratio",
        "recovered_norm_nt_distance_from_median",
    ]])

# Preserve the independently selected 3-month threshold result above, and
# also save a 3-month realization at the 24-month-selected norm_nt and seed.
companion_3m_from_24m = save_threshold_companion_exposure(
    threshold_results[24], exposure_months=3,
)
```

```python
for exposure_months in exposure_months_list:
    fig, axes = plot_norm_threshold_ensemble(threshold_results[exposure_months])
    fig.suptitle(f"{case['description']}, {exposure_months} months")
```

```python
sed_handoffs = {}
for exposure_months in exposure_months_list:
    sed_handoffs[exposure_months] = load_sed_handoff(threshold_results[exposure_months]["manifest_file"])
    print(exposure_months, sed_handoffs[exposure_months][0]["median_data_file"])

companion_handoff_3m = load_sed_handoff(
    companion_3m_from_24m["manifest_file"]
)
print("3-month companion at the 24-month norm_nt:", companion_handoff_3m[0]["median_data_file"])
```
